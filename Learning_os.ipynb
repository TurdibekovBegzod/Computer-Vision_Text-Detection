#### Fayl & Papka yaratish
- Fayl papka yaratishlik uchun bizda ikki hil metod bor
    1. <b>os.mkdir()</b> va <b>os.makedirs()</b>  funksiyalari
    2. <b>os.mkdir()</b> bosh bo'lgan bitta fayl papka yaratadi (ichma ich papka yaratish imkoni yo'q). Agar fayl mavjud bo'lsa <span style='color : red'>FileExistsError</span> xatoligini keltirib chiqaradi. Bunining oldini olishlik uchun esa <b>os.path.exists('<span style='color : lightblue'>path</span>')</b> (path ni o'rniga fayl joylashuvi ko'rsatilinadi)
    3. <b>os.makedirs()</b> ichma-ich bo'lgan papka yaratadi, agar mavjuda bo'lsa <span style='color : red'>FileExistsError</span> xatoligini keltirib chiqaradi. Bu xatolikni oldini olishlik uchun esa <b>exist_ok=True</b> giper-parametrni qo'shishlik kerak bo'ladi.

```python
os.mkdir("Hello")
```

```python
os.makedirs("Hello", exist_ok=True)
```

#### Fayl & Papkalarni o'chirish

- <b>remove()</b> va <b>removedirs()</b> o'chirish amallari bor 
- <b>remove()</b> faylni o'chiradi
- <b>removedirs()</b> bo'sh bo'lgan papkalarni o'chiradi, agarda ichma-ich papkalar bo'lsa va bo'sh bo'lsa o'chiriladi

```python
path = os.getcwd()
hello_path = os.path.join(path, 'Hello')
salom_path = os.path.join(path, 'salom.txt')
if os.path.exists(salom_path):
    os.remove(salom_path)


```

#### File nomlarini almashtirish

- os.rename(old_path, new_path)

```python
if os.path.exists(salom_path):
    os.rename('salom.txt', 'yana.txt')
```
